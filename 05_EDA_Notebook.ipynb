# Exploratory Data Analysis (EDA)
Analyzing Delays and Churn for SL Delivery Analytics Pro.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

orders = pd.read_csv('data/orders.csv')
customers = pd.read_csv('data/customers.csv')
orders['DelayMins'] = orders['ActualDeliveryMins'] - orders['ExpectedDeliveryMins']
```

## 1. Delay by Region

```
plt.figure(figsize=(10,6))
sns.barplot(data=orders, x='Region', y='DelayMins')
plt.title('Average Delay by Region')
plt.show()
```

## 2. Impact of Weather on Delays

```
plt.figure(figsize=(10,6))
sns.boxplot(data=orders, x='Weather', y='DelayMins')
plt.title('Weather Impact on Delays')
plt.show()
```

## 3. Churn vs Severe Delays

```
churn_rates = customers.groupby('SevereDelays')['Churn'].mean()
churn_rates.plot(kind='line', marker='o')
plt.title('Churn Probability vs Severe Delays Experienced')
plt.ylabel('Churn Rate')
plt.show()
```
